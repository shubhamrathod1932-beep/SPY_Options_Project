```python
import pandas as pd

calls = pd.read_csv("../data/options/SPY_Calls.csv")

calls.head()
```

```python
filtered_calls = calls[
    (calls["openInterest"] >= 100) &
    (calls["volume"] >= 10)
].copy()
```

```python
filtered_calls[
    [
        "strike",
        "lastPrice",
        "impliedVolatility",
        "volume",
        "openInterest"
    ]
].head(10)
```

```python
filtered_calls["IV %"] = filtered_calls["impliedVolatility"] * 100

filtered_calls[
    [
        "strike",
        "IV %"
    ]
].head(10)
```

```python
print(
    "Average IV:",
    filtered_calls["IV %"].mean()
)
```

```python
print(
    "Highest IV:",
    filtered_calls["IV %"].max()
)
```

```python
print(
    "Lowest IV:",
    filtered_calls["IV %"].min()
)
```

```python
filtered_calls.nlargest(
    5,
    "impliedVolatility"
)[
    [
        "strike",
        "IV %"
    ]
]
```

```python
filtered_calls.nsmallest(
    5,
    "impliedVolatility"
)[
    [
        "strike",
        "IV %"
    ]
]
```

```python
filtered_calls.to_csv("../data/options/SPY_IVfiltered_calls.csv", index=False)
```
