```python
vega = 0.25

iv_change = -12

price_change = vega * iv_change

print(price_change)
```

```python
vega_values = [0.10, 0.20, 0.30]

iv_drop = -15

for v in vega_values:
    print(
        f"Vega={v:.2f}",
        "Price Change=",
        round(v * iv_drop,2)
    )
```

```python
import matplotlib.pyplot as plt

events = [
    "Before Event",
    "After Event"
]

iv = [
    35,
    22
]

plt.figure(figsize=(6,5))

plt.bar(events, iv)

plt.ylabel("Implied Volatility (%)")

plt.title("Example of IV Crush")

plt.grid(axis="y")

plt.savefig(
    "../report/IV_Crush.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
