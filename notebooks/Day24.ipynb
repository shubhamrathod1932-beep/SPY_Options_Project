```python
import yfinance as yf
import pandas as pd
```

```python
spy = yf.Ticker("SPY")
```

```python
spy.options
```

```python
expiry = spy.options[0]

print(expiry)
```

```python
option_chain = spy.option_chain(expiry)
```

```python
calls = option_chain.calls

puts = option_chain.puts
```

```python
calls.head()
```

```python
puts.head()
```

```python
print(calls.columns)
```

```python
calls.to_csv("../data/options/SPY_Calls.csv", index=False)

puts.to_csv("../data/options/SPY_Puts.csv", index=False)

print("Option chain saved successfully!")
```
