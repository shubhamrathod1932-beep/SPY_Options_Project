```python
import pandas as pd

from src.greeks import (
    call_delta,
    gamma,
    vega,
    call_theta,
    call_rho
)

S = 727
T = 30/365
r = 0.04
sigma = 0.22

strikes = [700, 730, 760]

rows = []

for K in strikes:

    rows.append({
        "Strike": K,
        "Delta": call_delta(S,K,T,r,sigma),
        "Gamma": gamma(S,K,T,r,sigma),
        "Vega": vega(S,K,T,r,sigma),
        "Theta": call_theta(S,K,T,r,sigma),
        "Rho": call_rho(S,K,T,r,sigma)
    })

greeks_df = pd.DataFrame(rows)

greeks_df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.plot(
    greeks_df["Strike"],
    greeks_df["Delta"],
    marker="o"
)

plt.title("Delta vs Strike")

plt.xlabel("Strike")

plt.ylabel("Delta")

plt.grid(True)

plt.savefig(
    "../report/Delta_vs_Strike.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
