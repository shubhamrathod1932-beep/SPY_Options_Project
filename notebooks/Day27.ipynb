```python
import numpy as np
from scipy.stats import norm
```

```python
S = 730
K = 730
T = 30 / 365
r = 0.04
sigma = 0.20
```

```python
d1 = (
    np.log(S / K)
    + (r + sigma**2 / 2) * T
) / (sigma * np.sqrt(T))

d2 = d1 - sigma * np.sqrt(T)
```

```python
put_price = (
    K * np.exp(-r * T) * norm.cdf(-d2)
    - S * norm.cdf(-d1)
)

print(f"Theoretical Put Price = {put_price:.2f}")
```

```python
call_price = (
    S * norm.cdf(d1)
    - K * np.exp(-r * T) * norm.cdf(d2)
)

lhs = call_price + K * np.exp(-r * T)

rhs = put_price + S

print(lhs)
print(rhs)

print(lhs - rhs)
```

```python
def black_scholes_call(S, K, T, r, sigma):

    d1 = (
        np.log(S / K)
        + (r + sigma**2 / 2) * T
    ) / (sigma * np.sqrt(T))

    d2 = d1 - sigma * np.sqrt(T)

    return (
        S * norm.cdf(d1)
        - K * np.exp(-r * T) * norm.cdf(d2)
    )
```

```python
def black_scholes_put(S, K, T, r, sigma):

    d1 = (
        np.log(S / K)
        + (r + sigma**2 / 2) * T
    ) / (sigma * np.sqrt(T))

    d2 = d1 - sigma * np.sqrt(T)

    return (
        K * np.exp(-r * T) * norm.cdf(-d2)
        - S * norm.cdf(-d1)
    )
```

```python
print(black_scholes_call(730,730,30/365,0.04,0.20))

print(black_scholes_put(730,730,30/365,0.04,0.20))
```
