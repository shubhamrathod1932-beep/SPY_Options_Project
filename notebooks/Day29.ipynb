```python
import numpy as np
from scipy.stats import norm
```

```python
S = 730
K = 730
T = 30 / 365
r = 0.04
sigma = 0.20
```

```python
d1 = (
    np.log(S / K)
    + (r + sigma**2 / 2) * T
) / (sigma * np.sqrt(T))
```

```python
print("d1:", d1)
```

```python
call_delta = norm.cdf(d1)

put_delta = call_delta - 1

print("Call Delta =", call_delta)

print("Put Delta =", put_delta)
```

```python
def call_delta(S, K, T, r, sigma):

    d1 = (
        np.log(S / K)
        + (r + sigma**2 / 2) * T
    ) / (sigma * np.sqrt(T))

    return norm.cdf(d1)
```

```python
def put_delta(S, K, T, r, sigma):

    d1 = (
        np.log(S / K)
        + (r + sigma**2 / 2) * T
    ) / (sigma * np.sqrt(T))

    return norm.cdf(d1) - 1
```

```python
print(call_delta(730,730,30/365,0.04,0.20))

print(put_delta(730,730,30/365,0.04,0.20))
```
