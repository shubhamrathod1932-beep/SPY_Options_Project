```python
import numpy as np
from scipy.stats import norm
```

```python
S = 730
K = 730
T = 30 / 365
r = 0.04
sigma = 0.20
```

```python
d1 = (
    np.log(S / K)
    + (r + sigma**2 / 2) * T
) / (sigma * np.sqrt(T))

print(d1)
```

```python
d2 = d1 - sigma * np.sqrt(T)

print(d2)
```

```python
print(norm.cdf(d1))
print(norm.cdf(d2))
```

```python
discounted_strike = K * np.exp(-r * T)

print(discounted_strike)
```

```python
call_price = (
    S * norm.cdf(d1)
    - discounted_strike * norm.cdf(d2)
)

print(call_price)
```
