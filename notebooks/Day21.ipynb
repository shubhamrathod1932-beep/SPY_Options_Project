```python
import pandas as pd
import yfinance as yf

# Download 1 year of SPY data
spy = yf.download("SPY", period="1y")

# Display first 5 rows
print(spy.head())

# Display information about the DataFrame
print("\nData Types:")
print(spy.dtypes)

print("\nSummary Statistics:")
print(spy.describe())

# Save the data
spy.to_csv("../data/historical/SPY_1Y.csv", index=False)

print("\nSPY data saved successfully!")
```
