```python
import numpy as np
import pandas as pd
from scipy.stats import norm
```

```python
calls = pd.read_csv("../data/options/SPY_Calls.csv")
```

```python
option = calls.iloc[111]

print(option)
```

```python
S = 738.25
K = option["strike"]

sigma = option["impliedVolatility"]

market_price = option["lastPrice"]

T = 1 / 365

r = 0.04
```

```python
def black_scholes_call(S, K, T, r, sigma):

    d1 = (
        np.log(S / K)
        + (r + sigma**2 / 2) * T
    ) / (sigma * np.sqrt(T))

    d2 = d1 - sigma * np.sqrt(T)

    return (
        S * norm.cdf(d1)
        - K * np.exp(-r * T) * norm.cdf(d2)
    )
```

```python
theoretical_price = black_scholes_call(
    S,
    K,
    T,
    r,
    sigma
)

print(theoretical_price)
```

```python
pricing_error = market_price - theoretical_price

print(pricing_error)
```

```python
if pricing_error > 0:
    print("Market price is above theoretical price.")

elif pricing_error < 0:
    print("Market price is below theoretical price.")

else:
    print("Market price equals theoretical price.")
```
