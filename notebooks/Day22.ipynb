```python
import pandas as pd
import numpy as np

# Load SPY data
df = pd.read_csv("../data/historical/SPY_1Y.csv")

# If needed, flatten MultiIndex column names
if isinstance(df.columns, pd.MultiIndex):
    df.columns = [col[0] for col in df.columns]

# Calculate returns
df["Simple Return"] = df["Close"].pct_change()
df["Log Return"] = np.log(df["Close"] / df["Close"].shift(1))

# Display first 10 rows
print(df[["Close", "Simple Return", "Log Return"]].head(10))

# Remove NaN values
df = df.dropna()

# Summary statistics
print("\nSummary Statistics:")
print(df[["Simple Return", "Log Return"]].describe())

# Save the updated data
df.to_csv("../data/historical/SPY_Returns.csv", index=False)

print("\nSPY returns saved successfully!")
```
