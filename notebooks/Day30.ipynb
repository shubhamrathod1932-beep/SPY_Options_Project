```python
import numpy as np
from scipy.stats import norm
```

```python
S = 730
K = 730
T = 30 / 365
r = 0.04
sigma = 0.20
```

```python
d1 = (
    np.log(S / K)
    + (r + sigma**2 / 2) * T
) / (sigma * np.sqrt(T))
```

```python
print("d1:", d1)
```

```python
gamma = norm.pdf(d1) / (
    S * sigma * np.sqrt(T)
)

print("Gamma =", gamma)
```

```python
def gamma(S, K, T, r, sigma):

    d1 = (
        np.log(S / K)
        + (r + sigma**2 / 2) * T
    ) / (sigma * np.sqrt(T))

    return norm.pdf(d1) / (
        S * sigma * np.sqrt(T)
    )
```

```python
print(gamma(730, 730, 30/365, 0.04, 0.20))
```

```python
print(gamma(730, 700, 30/365, 0.04, 0.20))
```

```python
print(gamma(730, 760, 30/365, 0.04, 0.20))
```
